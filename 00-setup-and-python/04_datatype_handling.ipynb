```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "age": [21, 25, 30, 35],
    "salary": [25000.50, 32000.00, 45000.75, 61000.00],
    "city": ["Delhi", "Mumbai", "Delhi", "Pune"],
    "subscribed": [True, False, True, False],
    "join_date": [
        "2024-01-10",
        "2024-03-15",
        "2024-06-20",
        "2024-09-05"
    ]
})
```

```python
print(df)
```

```python
print(df.dtypes)
```

```python
df.info()
```

```python
print(df.shape)
```

```python
df["join_date"] = pd.to_datetime(df["join_date"])

print(df.dtypes)
```

```python
df["join_year"] = df["join_date"].dt.year
df["join_month"] = df["join_date"].dt.month

print(df[["join_date", "join_year", "join_month"]])
```

```python
df["salary"] = df["salary"].astype(str)

print(df["salary"].dtype)
```

```python
df["salary"] = pd.to_numeric(
    df["salary"],
    errors="coerce"
)

print(df["salary"].dtype)
```
