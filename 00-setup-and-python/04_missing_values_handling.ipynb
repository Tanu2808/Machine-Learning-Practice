```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "age": [20, 21, 19, np.nan, 22],
    "study_hours": [3, 5, np.nan, 2, 4],
    "exam_score": [75, 88, 70, 60, np.nan]
})
```

```python
print(df)
```

```python
# Number of missing values in each column
print(df.isna().sum())
```

```python
print(df.isna().sum().sum())
```

```python
print(df.isna().mean() * 100)
```

```python
df_filled = df.copy()
```

```python
df_filled = df.copy()

df_filled["age"] = df_filled["age"].fillna(
    df_filled["age"].median()
)

df_filled["study_hours"] = df_filled["study_hours"].fillna(
    df_filled["study_hours"].median()
)

df_filled["exam_score"] = df_filled["exam_score"].fillna(
    df_filled["exam_score"].median()
)

print(df_filled)
```

```python
print(df_filled.isna().sum())
```
