```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "city": ["Delhi", "Mumbai", "Pune", "Delhi", "Mumbai"],
    "membership": [
        "Bronze", "Silver", "Gold", "Gold", "Bronze"
    ],
    "satisfaction": [
        "Low", "Medium", "High", "Medium", "Low"
    ],
    "purchased": ["Yes", "No", "Yes", "Yes", "No"]
})
```

```python
print(df)
```

```python
print(df.dtypes)
print(df["city"].unique())
print(df["city"].value_counts())
```

```python
df_onehot = pd.get_dummies(
    df,
    columns=["city"],
    dtype=int
)

print(df_onehot)
```

```python
from sklearn.preprocessing import OrdinalEncoder

order = [["Low", "Medium", "High"]]

encoder = OrdinalEncoder(
    categories=order
)

df["satisfaction_encoded"] = encoder.fit_transform(
    df[["satisfaction"]]
).astype(int)

print(df[["satisfaction", "satisfaction_encoded"]])
```

```python
df["purchased_encoded"] = (
    df["purchased"].map({"No": 0, "Yes": 1})
)

print(df[["purchased", "purchased_encoded"]])
```

```python
print(df["purchased_encoded"].isna().sum())
```
