```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "transaction_id": [101, 102, 103, 103, 104, 105],
    "customer": [
        "Amit", "Priya", "Rahul",
        "Rahul", "Neha", "Amit"
    ],
    "product": [
        "Laptop", "Phone", "Mouse",
        "Mouse", "Laptop", "Laptop"
    ],
    "amount": [50000, 20000, 800, 800, 48000, 50000]
})
```

```python

print(df)
```

```python
print(df.duplicated().sum())
```

```python
duplicates = df[df.duplicated()]

print(duplicates)
```

```python
duplicates_all = df[
    df.duplicated(keep=False)
]

print(duplicates_all)
```

```python
print(
    df[df.duplicated(
        subset=["transaction_id"],
        keep=False
    )]
)
```

```python
print(
    df[df.duplicated(
        subset=["customer", "product"],
        keep=False
    )]
)
```

```python
df_clean = df.drop_duplicates()

print(df_clean)
```

```python
print("Rows before:", len(df))
print("Rows after:", len(df_clean))
print("Remaining exact duplicates:", df_clean.duplicated().sum())
```
